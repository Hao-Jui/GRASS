```python
# FUKA initial data APR, if you want to change EOS, you just need to change the dictionary name
logp1, Gamma1, Gamma2,Gamma3 = EOS_Read_et_al_dict['15H']
eos=EOS4(logp1, Gamma1, Gamma2, Gamma3)
n = 2000
# density change from 1e-3 to 1e15.5
index = np.linspace(-3, 15.5, n)
rho_list = 10 ** index

with open('15H.lorene', 'w') as file:

    file.write("#\n")
    file.write("#\n")
    file.write("#\n")
    file.write("#\n")
    file.write("#\n")
    file.write(f"{n}\n")
    file.write("#\n")
    file.write("# index  n_B [fm^{-3}]  rho [g/cm^3]  p [dyn/cm^2]\n")
    file.write("#\n")

    for i in range(n):
        e = eos.rho2e(rho_list[i])
        p = eos.rho2p(rho_list[i])
        nb = rho_list[i]/mu/1.0e39
        file.write(f"{i+1:8}\t{nb:24.16e}\t{e:24.16e}\t{nb:24.16e}\n")
```

```python
# SACRA EOS 

logp1, Gamma1, Gamma2,Gamma3 = EOS_Read_et_al_dict['H4']
eos=EOS4(logp1, Gamma1, Gamma2, Gamma3)
n = 2000
index = np.linspace(-3, 15.5, n)
rho_list = 10.0 ** index

with open('../H4.dat', 'w') as file:

    for i in range(n):
        e = eos.rho2e(rho_list[i])
        p = eos.rho2p(rho_list[i])
        nb = rho_list[i]/mu/1.0e39
        eps = e/rho_list[i]-1
        hh = (e+p/C/C)/rho_list[i]
        dpdrho = eos.rho2gamma(rho_list[i])*p/rho_list[i]
        cs = eos.rho2gamma(rho_list[i]) * p /(p+e*C**2)
        file.write(f"{e:.16e}\t{p:.16e}\t{hh:.16e}\t{rho_list[i]/1.6749286e-24:.16e}\n")
```
